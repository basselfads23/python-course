```python
import numpy as np
import matplotlib.pyplot as plt

data=np.loadtxt('64-17-5-IR.jdx',unpack=True, skiprows=25)
x=[];y=[]
for i in range(0,len(data[0])):
    x.append(data[0][i])
    for j in range(1,10):
        x.append(data[0][i]+4*j)
    for j in range(1,11):
        y.append(data[j][i])

x = np.array(x)
y = np.array(y)

plt.plot(x,y)
```

```python
xi2 = np.where( x >= 3502)[0][0]
x2 = x[xi2:]
y2 = y[xi2:]
```

```python
plt.plot(x2,y2)
```

```python
def Lorentzian(x, A,gam,x0):
    num = 2*A/(np.pi*gam)
    deno = 1 + (2*(x-x0)/gam)**2
    return num / deno
```

```python
ymax = np.max(y2)
x0 = np.where( x >= 3502)[0][0]
```

```python
from scipy.optimize import curve_fit

p0 = np.array([ymax,1,x0])

p , _ = curve_fit(Lorentzian, x2, y2, p0=p0)

y_opti = Lorentzian(x2, p[0],p[1],p[2])

plt.plot(x2,y2, '.')
plt.plot(x2,y_opti)
```
