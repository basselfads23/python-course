```python
import time 
import serial 
import serial.tools.list_ports as port_list 
```

```python
ports = list(port_list.comports())
for p in ports: print(p)
```

```python
serial_add = 'the port detected'
ser = serial.Serial(serial_add,baudrate=9600,bytesize=7, parity='E',stopbits=1,timeout=.5)
```

```python
ser.write(b"i:30\r\n")   
l = ser.readline()
response_state=l.decode("ascii", errors="replace")
print(response_state)
```

```python
ser.write(b"A:\r\n") 
response_position = ser.readline().decode("ascii").strip()
print(response_position)
```

```python
position = 5000 
time_hold = 120 

if 0 <= position <= 100000:
    cmd = f"R:{position:06d}\r\n" 
    ser.write(cmd.encode("ascii")) 

    pos = 100002  
    while abs(pos - position) >= 5 : 
        ser.write(b"A:\r\n")
        line = ser.readline().decode("ascii").strip()
        pos = int(line[2:])
        print(f"Position of the valve : {pos}", end='\r')

        time.sleep(1)
        if pos == 999999 :
            print("ERROR")
            break
    
    time.sleep(time_hold)  
    
    
    ser.write(b"C:\r\n")
else:
    print("Error : position value must be between 0 and 100000.")
```
